```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler , OneHotEncoder

df = pd.DataFrame({
    "hours_studied":[2,4,6,8,10],
    "study_method":["book", "video", "online","video", "book"]

})
print("BEFORE TRANSORMATION")
print(df)
prep = ColumnTransformer([
    ("num", StandardScaler(), ["hours_studied"]),
    ("cat", OneHotEncoder(handle_unknown="ignore"), ["study_method"])
])
transformed_data = prep.fit_transform(df)
columns = prep.get_feature_names_out()
transformed_df = pd.DataFrame(transformed_data, columns=columns)
print("\nAFTER TRANSFORMATION")
print(transformed_df.round(2))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
prep = ColumnTransformer([
    ("num", StandardScaler(), ["hours_studied"]),
    ("cat", OneHotEncoder(handle_unknown="ignore"), ["study_method"])
])
pipe = Pipeline([
    ("preprocess", prep),
    ("model", LogisticRegression(max_iter=1000))
])
pipe.fit(X_train, y_train)
#print(pipe.predict(X_test))
#print(pipe.score(X_test, y_test))
#print(pipe.named_steps['model'].n_iter_)
```
